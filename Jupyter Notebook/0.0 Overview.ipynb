# Overview
This Jupyter notebook serves as documentation for the Amplify library.  The intended use of Amplify is to simplify the formulation and solution of reproducible robotics optimization problems (ROPs) that are expressed as NLPs in the AMPL modeling language.  We only cover syntax related to Amplify and AMPL.  We assume the reader is familiar with python and Jupyter notebooks.


## An AMPL Primer
A native AMPL implementation breaks the library into three components:
1. A model file where the library and user code reside,
2. A user-defined data file where values of parameters, variables, and sets reside, and
3. A user-defined run file, where the problem set up and solution occur.

In this notebook, the three files are intermingled to help make the experience more interactive.  In general, code prefixed with a `data;` command are meant to go into an AMPL data (.dat) file.  The next code cell and the final two code cells are meant to go into the run (.run) file.  All other code cells are part of the model (.mod) file.

### Learning the Language
The suggested reference for new AMPL users is [AMPL: A Modeling Language for Mathematical Programming](https://ampl.com/resources/ampl-book/).  Other references are available on the [AMPL website](https://ampl.com/resources/#learnbuild).

**TIP:** Do you need to read the book before proceeding?  In general, if you've coded in enough programming languages and are familiar with how NLPs are formulated on paper, it should be possible to guess what most of the Amplify code is doing.  If in doubt, the AMPL book is a click away (and freely available).


## The Rest of the Content
The remaining notebooks consists of all of the modules of the Amplify library and a running example of finding optimal walking gaits for a five-link biped robot.  The modules are the
* Grid,
* Bézier,
* State,
* Robot Kinematic Tree,
* Robot Constraints,
* Spatial Quantities,
* Composite Rigid Body,
* Recursive Newton-Euler,
* Task Jacobian,
* Equations of Motion, and
* Runge-Kutta modules.

In AMPL model and data files, each of these modules are self-contained blocks of code.  They parse their own inputs, add constraints, and define sets, parameters, and decision variables for other modules to use.

**TIP:** Don't like a module definition, feel free to change it!  Any module `M` in Amplify can be replaced as long as the new module defines any sets, parameters, and decision variables that dependent modules expect `M` to define.  For example, the `Recursive Newton-Euler` module defines `var b` which the `Equations of Motion` module depends on.  Any new module that is meant to replace the `Recursive Newton-Euler` module is obligated to also define `b` so that the overall library still functions.
